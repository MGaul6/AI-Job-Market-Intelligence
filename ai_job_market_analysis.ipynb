```python
df['posted_date'] = pd.to_datetime(df['posted_date'])
```

```python
df[['min_salary','max_salary']] = df['salary_range_usd'].str.split('-', expand=True)
```

```python
df[['salary_range_usd','min_salary','max_salary']].head()
```

```python
df['min_salary'] = pd.to_numeric(df['min_salary'])
df['max_salary'] = pd.to_numeric(df['max_salary'])
```

```python
df['avg_salary'] = (df['min_salary'] + df['max_salary']) / 2
```

```python
df[['min_salary','max_salary','avg_salary']].head()
```

Basic Salary Analysis

```python
df['avg_salary'].describe()
```

Highest-Paying Roles

```python
df.groupby('job_title')['avg_salary'].mean().sort_values(ascending=False).head(10)
```

```python
df['job_title'].value_counts().head(10)
```

**Salary by Experience**

```python
df.groupby('experience_level')['avg_salary'].mean().sort_values(ascending=False)
```
